```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
height = np.random.normal(140 , 10 , 700)
# height
```

```python
plt.figure()
plt.plot(height , linestyle = "None" , marker = "o" , color  = "red")
plt.axhline((np.mean(height) + np.std(height)) , color = "blue" , label = "std dev")
plt.axhline((np.mean(height) - np.std(height)) , color = "blue" , label = "std dev")
plt.axhline((np.var(height) + np.mean(height)))
plt.axhline((np.var(height) - np.mean(height)))

plt.axhline(np.mean(height) , color = "green" , label = "mean")
plt.axhline(np.median(height) , color  = "black" , label = "median")
plt.legend()
plt.show()
```

```python
plt.figure()
plt.hist(height , color = "red" , ec = "black"  , bins=30)
plt.axvline(np.mean(height) , color = "blue")
plt.axvline(np.median(height) , color = "blue")

plt.show()
```

```python
print(np.mean(height) , np.median(height) , sep = "\n")
```
