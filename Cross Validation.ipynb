```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = sns.load_dataset("mpg")
data
```

```python
data.dropna(inplace = True)  # dropping missing values
```

```python
x = data.drop("mpg" , axis = 1)
y = data["mpg"]
```

```python
x
```

```python
y
```

```python
x.drop(["origin" , "name"] , axis = 1 , inplace = True)
x
```

```python
for i in x :
    plt.figure()
    plt.scatter(x[i] , y , color = "red")
    plt.xlabel(i)
    plt.ylabel("MPG")
    plt.title(f"{i} vs MPG")
    plt.show()
    print("-"*50)
```

```python
x.corrwith(y)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score , mean_absolute_error
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2)
```

```python
model = LinearRegression()
model
```

```python
model.fit(x_train , y_train)
y_pred = model.predict(x_test)
print(f"Accuracy --> {r2_score(y_test , y_pred)}")
print(f"Error --> {mean_absolute_error(y_test , y_pred)}")
```

```python
height = np.random.normal(150 , 15 , 100)
weight = 0.6 * height + np.random.rand(100) * 30
```

```python
plt.scatter(height , weight)
plt.show()
```

```python
Poly = PolynomialFeatures(include_bias = False)
height_new = Poly.fit_transform(height.reshape(-1 , 1))
height_new
```

```python
model = LinearRegression()
model.fit(height.reshape(-1,1) , weight)
y_pred = model.predict(height.reshape(-1,1))
print(f"Accuracy --> {r2_score(weight , y_pred)}")
```

```python
model = LinearRegression()
model.fit(height_new , weight)
y_pred = model.predict(height_new)
print(f"Accuracy --> {r2_score(weight , y_pred)}")
```

```python
x_train_new = Poly.fit_transform(x_train)
x_test_new = Poly.fit_transform(x_test)
```

```python
model = LinearRegression()
model.fit(x_train_new , y_train)
y_pred = model.predict(x_test_new)
print(f"Accuracy --> {r2_score(y_test , y_pred)}")
```
