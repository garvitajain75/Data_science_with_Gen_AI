```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
height = np.random.normal(140,10,800)
```

```python
plt.figure()
plt.plot(height, linestyle = "None" , marker = "o" , color = "red")
plt.axhline(np.mean(height) , label = "mean")
plt.axhline(np.median(height) , color = "green" , label = "median")
plt.legend()
plt.show()
```

```python
plt.figure()
plt.plot(height , linestyle = "None" , marker = "o" , color  = "red")
plt.axhline((np.mean(height) + np.std(height)) , color = "blue" , label = "std dev")
plt.axhline((np.mean(height) - np.std(height)) , color = "blue" , label = "std dev")
plt.axhline((np.mean(height) + np.var(height)) , color = "yellow" , label = "std dev")
plt.axhline((np.mean(height) - np.var(height)) , color = "yellow" , label = "std dev")

plt.axhline(np.mean(height) , color = "green" , label = "mean")
plt.axhline(np.median(height) , color  = "black" , label = "median")
plt.legend()
plt.show()
```
