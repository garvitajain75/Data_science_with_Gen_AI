# What is Pandas ?


- Pandas is an open-source Python library for data manipulation, cleaning, analysis, and visualization.

- It is built on Numpy, which means it provides fasst vectorized operations.

- Pandas is optimized for tabular data (like Excel sheets, SQL tables, CSV files).

# Key Terms
- Multivariate  (Univariate -- Normal distribution , Bivariate -- if Age increase, weight is also increased)
- Data wrangling  (a part of Data Manipulation)

# Key featurs:
- Fast and efficient DataFrame object for data manipulation.
- Handling of missing data.
- Flexible reshaping and pivoting of datasets.
- Label-based slicing, indexing, and subsetting.
- Merging and joining of datasets.
- Time series functionality
- Integration with visualization libraries like Matpolib.
