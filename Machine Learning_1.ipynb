```python
data = pd.read_csv("housing 1.csv")

x = data.drop("median_house_value" ,axis = 1)
y = data["median_house_value"]

num_col_list = x.select_dtypes(["int" , "float"]).columns.tolist()
cat_col_list = x.select_dtypes("object").columns.tolist()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer # impute means missing values ko fill krna
```

```python
preprocessor = ColumnTransformer(
    transformers = [
        ("Encoding" , OneHotEncoder(sparse_output = False) , cat_col_list) ,
        ("Impute" , SimpleImputer(strategy = "median") , num_col_list)        # OneHot Encoder apply krna h cat_col_list p
    ]
)
preprocessor
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size = 0.2)
```

```python
preprocessor.fit(x_train)
```

```python
x_train_prep = preprocessor.transform(x_train)
x_test_prep = preprocessor.transform(x_test)
```

```python
x_train_prep
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model
```

```python
model.fit(x_train_prep , y_train) # estimation / learning
```

```python
y_pred = model.predict(x_test_prep)
y_pred
```

```python
from sklearn.metrics import r2_score
r2_score(y_test , y_pred) * 100
```
