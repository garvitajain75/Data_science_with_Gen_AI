```python
z = np.array([200,400,500,900,100])
z
```

```python
jaipur = np.random.normal(35,5,7).astype(int)
jaipur
```

```python
noida = np.random.normal(35,5,7).astype(int)
noida
```

```python
alwar = np.random.normal(35,5,7).astype(int)
alwar
```

```python
gurugram = np.random.normal(35,5,7).astype(int)
gurugram
```

```python
fig = plt.figure(facecolor = "green" , edgecolor = "blue")
fig.set_linewidth(5)
plt.plot(x , jaipur , color = "red" , mfc = "green" , mec = "blue" , linestyle = "-." , label = "Jaipur")
plt.plot(x , noida , color = "yellow" , mfc = "green" , mec = "blue" , linestyle = "-." , label = "Noida")
plt.plot(x , alwar  , color = "green" , mfc = "green" , mec = "blue" , linestyle = "-." , label = "Alwar")
plt.plot(x , gurugram  , color = "blue" , mfc = "green" , mec = "blue" , linestyle = "-." , label = "Gurugram")
plt.legend()
plt.gca().set_facecolor("lightblue")

plt.title("Temperature Chart" , color = "blue", fontsize = 15)
plt.xlabel("Days" , color = "White")
plt.ylabel("Temperature" , color = "White")
plt.grid(color = "lightblue" , linestyle = "--" , axis = "x")
plt.tick_params(axis = "y" , direction = "inout" , length = 14 , width = 4 , pad = 8 , labelcolor = "b" , color = "green")  
plt.show()
```
