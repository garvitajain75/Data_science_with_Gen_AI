# Merge

```python
# pd.merge()
```

```python
d1 = pd.DataFrame({
    "marks" : np.random.normal(80 , 10 , 22) ,
    "age" : np.random.normal(20 , 4 , 22) ,
    "roll" : np.arange(1,23)
})
d1
```

```python
d2 = pd.DataFrame({
    "physics_score" : np.random.normal(80 , 10 , 20) ,
    "avg_age" : np.random.normal(20 , 4 , 20) ,
    "roll" : np.arange(1,21)
})
d2
```

```python
pd.merge(d1 , d2 , on = "roll")  # by default inner join
```
