```python
data.groupby("sex")["total_bill"].mean()
```

```python
data[data["sex"] == "Male"]["total_bill"].mean()
```

```python
data[data["sex"] == "Female"]["total_bill"].mean()
```

```python
data.groupby("day")["total_bill"].mean()
```

```python
data.groupby("day")[["total_bill","tip"]].mean()
```

```python
data.groupby("sex")[["total_bill" , "tip"]].agg(["mean" , "median" , "sum"])
```

```python
data.groupby("day")[["total_bill","tip"]].agg({
    "total_bill" : "mean" ,
    "tip" : "median"
})
```

```python
data.groupby("day")[["total_bill","tip"]].agg({
    "total_bill" : ["mean" , "median"] ,
    "tip" : ["median" , "sum"]
})
```

```python
df = sns.load_dataset("flights")
df
```
