```python
# import required libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv("housing 1.csv")
data
```

```python
data.dropna(inplace = True)
```

```python
x = data.drop("median_house_value" , axis = 1)   # independent variable / feature
y = data["median_house_value"]  # dependent var / target
```

```python
cat_col_list = x.select_dtypes("object").columns.tolist()
num_col_list = x.select_dtypes(["int" , "float"]).columns.tolist()
print(cat_col_list , num_col_list , sep = "\n")
```

```python
preprocessor = ColumnTransformer(
    transformers = [
        ("Scale" , StandardScaler() , num_col_list) ,
        ("Encoder" , OneHotEncoder(sparse_output=False) , cat_col_list) 
    ] 
)
preprocessor
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2)
```

```python
x_train_scaled = preprocessor.fit_transform(x_train) # training data
x_test_scaled = preprocessor.transform(x_test)
```

```python
model = LinearRegression()
model
```

```python
model.fit(x_train_scaled , y_train)  # estimation ....
y_pred = model.predict(x_test_scaled)
print(f"Testing Accuracy --> {r2_score(y_test , y_pred) * 100}")
print(f"Testing Error --> {mean_absolute_error(y_test , y_pred)}")
```

# Data Pipeline

```python
num_col_list
```

```python
cat_col_list
```

```python
from sklearn.pipeline import Pipeline
```

```python
Num_Pipeline = Pipeline(
    steps = [
        ("Impute" , SimpleImputer(strategy="median")) ,
        ("Scale" , StandardScaler())
    ]
)
Num_Pipeline
```

```python
cat_pipeline = Pipeline(
    steps = [
        ("Impute" , SimpleImputer(strategy = "most_frequent")) ,
        ("Encode" , OneHotEncoder(sparse_output = False))
    ]
)
cat_pipeline
```

```python
ct = ColumnTransformer(
    transformers = [
        ("num_process" , Num_Pipeline , num_col_list) ,
        ("Cat_process" , cat_pipeline , cat_col_list)
    ]
)
ct
```

```python
x_train_process = ct.fit_transform(x_train)
x_test_process = ct.transform(x_test)
```

```python
final_Pipeline = Pipeline(
    steps = [
        ("preprocess" , ct) ,
        ("model" , LinearRegression())
    ]
)
final_Pipeline
```

```python
final_Pipeline.fit(x_train,y_train)
```

```python
y_pred = final_Pipeline.predict(x_test)
```

```python
r2_score(y_test , y_pred)
```
